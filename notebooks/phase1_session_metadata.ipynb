```python
%load_ext autoreload
%autoreload 2

import sys
sys.path.append("../")

from tools.behavior import load_session_registry
```

## Load Session Registry

```python
# Load the session registry
registry = load_session_registry()
print(registry)
```

## View All Sessions

```python
# Print all sessions
for session in registry.sessions:
    print(f"{session.session_name:25s} | {session.animal_id:4s} | {session.condition:8s} | {session.label}")
```

## Filter by Animal

```python
# Get all sessions for M103
m103_sessions = registry.filter(animal_ids=["M062"])
print(f"M103 Sessions ({len(m103_sessions)} total):")
for s in m103_sessions:
    print(f"  {s.short_label}: {s.date.date()}")
```

## Filter by Condition

```python
# Get all muscimol sessions
muscimol_sessions = registry.filter(conditions=["control"])
print(f"Muscimol Sessions ({len(muscimol_sessions)} total):")
for s in muscimol_sessions:
    print(f"  {s.label}: {s.date.date()}")
```

## Filter by Animal AND Condition

```python
# Get all normal sessions for animals M103 and M106
normal_m103_m106 = registry.filter(
    animal_ids=["M103", "M106"],
    conditions=["normal"]
)
print(f"Normal sessions for M103 & M106:")
for s in normal_m103_m106:
    print(f"  {s.label}")
```

## Get Specific Session

```python
# Look up a specific session by name
session = registry.get("M103_2026_02_17_14_00")
print(f"Session: {session.session_name}")
print(f"  Animal: {session.animal_id}")
print(f"  Condition: {session.condition}")
print(f"  Date: {session.date}")
print(f"  Day: {session.day_num}")
print(f"  Label: {session.label}")
print(f"  Short label: {session.short_label}")
```

## Summary Statistics

```python
# Get unique animals and conditions
print(f"Animals: {registry.animals()}")
print(f"Conditions: {registry.conditions()}")

# Count sessions per condition
print("\nSessions per condition:")
for cond in registry.conditions():
    count = len(registry.filter(conditions=[cond]))
    print(f"  {cond}: {count}")

# Count sessions per animal
print("\nSessions per animal:")
for animal in registry.animals():
    count = len(registry.filter(animal_ids=[animal]))
    print(f"  {animal}: {count}")
```
